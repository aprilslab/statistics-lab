불공정한 룰렛의 확률 밀도함수
$$
f(x) = \left\{
\begin{array}{ll}
2x && (0 \le x \le 1) \\
0 && (otherwise)
\end{array}
\right.
$$


$$
P(0.4 \le X \le 0.6) = \int_{0.4}^{0.6} 2x \, dx
$$

```python
def f(x):
    if 0 <= x <= 1:
        return 2 * x
    else:
        return 0
```

```python
# 시각화
```

#### 기댓값

$$
\mu = E(X) = \int_{-\infty}^{\infty} xf(x)dx
$$

#### 분산
$$
\sigma^2 = V(X) = \int_{-\infty}^{\infty}(x-\mu)^2f(x)dx
$$
